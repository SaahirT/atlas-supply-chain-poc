# ATLAS - Traffic Data Cleaning

## Purpose

For this notebook, the goal is to clean the US Accidents dataset and create a smaller dataset containing only traffic incidents located in New Jersey.

The cleaned traffic data will later be used to identify transportation risk patterns in the ATLAS dashboard.

## Step 1: Import Libraries

The first step is to import the Python libraries needed for this notebook.

Pandas will be used to load, clean, filter, and transform the traffic dataset.

```python
import pandas as pd
```

## Step 2: Load the Traffic Dataset

The US Accidents dataset comes from Kaggle and contains historical traffic incident data collected from transportation and traffic sources across the United States.

This dataset contains information about traffic incidents, including their location, severity, timing, roadway impact, and weather conditions.

The raw dataset is stored in the `data/raw` folder.

```python
traffic_df = pd.read_csv("../data/raw/US_Accidents_March23.csv")
```

## Step 3: Inspect the Dataset

Before cleaning the data, it is important to understand what the dataset looks like.

The following cells will display:

- The first few rows
- The number of rows and columns
- The names of all the columns

```python
traffic_df.head()
```

```python
traffic_df.shape
```

```python
traffic_df.columns.tolist()
```

The dataset has 7,728,394 rows (each representing a traffic incident), and 46 columns providing different information about that incident.

## Step 4: Filter for New Jersey Incidents

Since this is a POC for J&J, ATLAS will focus only on logistics within New Jersey.

So, the full US Accidents dataset will be filtered to only traffic incidents where the state is listed as `NJ`.

```python
nj_traffic = traffic_df[traffic_df["State"] == "NJ"].copy()

nj_traffic.shape
```

```python
nj_traffic.head()
```

After filtering the dataset, we are left with 140,719 incidents that are based in New Jersey.

## Step 5: Select Relevant Columns

For this project, only the columns needed to understand traffic incidents and their surrounding conditions will be kept:

- ID
- Severity
- Start Time
- End Time
- Start Latitude
- Start Longitude
- City
- County
- State
- ZIP Code
- Distance (mi)
- Temperature (F)
- Visibility (mi)
- Precipitation (in)
- Weather Condition

```python
traffic_columns = [
    "ID",
    "Severity",
    "Start_Time",
    "End_Time",
    "Start_Lat",
    "Start_Lng",
    "City",
    "County",
    "State",
    "Zipcode",
    "Distance(mi)",
    "Temperature(F)",
    "Visibility(mi)",
    "Precipitation(in)",
    "Weather_Condition"
]

nj_traffic = nj_traffic[traffic_columns]

nj_traffic.shape
```

```python
nj_traffic.head()
```

## Step 6: Check for Missing Values

Next we will check for missing values which could cause problems later when the data is used for traffic-risk and geographic analysis.

```python
nj_traffic.isnull().sum()
```

```python
nj_traffic = nj_traffic.dropna(subset=["Zipcode"])

nj_traffic.isnull().sum()
```

The core traffic fields needed for ATLAS, including incident ID, severity, timing, and location, contain no missing values.

One incident is missing a ZIP code, so that row will be removed.

Some weather-related fields contain missing values. These rows will be kept because the traffic incident itself is still useful, even if some weather information is unavailable.

## Step 7: Create Time-Based Fields

The traffic dataset includes start and end timestamps for each incident.

These timestamps will be converted into a datetime format so ATLAS can analyze traffic patterns by:

- Hour of day
- Day of week
- Incident duration

```python
nj_traffic["Start_Time"] = pd.to_datetime(
    nj_traffic["Start_Time"],
    format="mixed"
)

nj_traffic["End_Time"] = pd.to_datetime(
    nj_traffic["End_Time"],
    format="mixed"
)

nj_traffic["Hour"] = nj_traffic["Start_Time"].dt.hour
nj_traffic["Day_of_Week"] = nj_traffic["Start_Time"].dt.day_name()

nj_traffic["Incident_Duration_Min"] = (
    nj_traffic["End_Time"] - nj_traffic["Start_Time"]
).dt.total_seconds() / 60

nj_traffic[
    ["Start_Time", "End_Time", "Hour", "Day_of_Week", "Incident_Duration_Min"]
].head()
```

## Step 8: Validate Incident Duration

Before using incident duration in later analysis, the values will be reviewed for possible data issues.

This check will look for:

- Negative durations
- Extremely long durations
- General summary statistics

```python
nj_traffic["Incident_Duration_Min"].describe()
```

```python
(nj_traffic["Incident_Duration_Min"] < 0).sum()
```

```python
nj_traffic.nlargest(
    10,
    "Incident_Duration_Min"
)[
    ["ID", "Start_Time", "End_Time", "Incident_Duration_Min"]
]
```

No negative incident durations were found.

Most incidents have reasonable durations, with a median duration of about 84 minutes. However, a small number of records contain extremely long durations lasting months or even years.

These extreme values are likely not useful for normal traffic-risk analysis. The incidents will remain in the dataset, but unusually long durations will be excluded when incident duration is analyzed.

## Step 9: Identify Valid Incident Durations

Some traffic incidents contain unusually long duration values.

For ATLAS, incident durations of 24 hours or less will be considered reasonable for normal traffic-risk analysis.

Longer incidents will remain in the dataset, but their duration values will not be used when analyzing typical incident duration.

```python
nj_traffic["Valid_Duration"] = (
    nj_traffic["Incident_Duration_Min"] <= 1440
)

nj_traffic["Valid_Duration"].value_counts()
```

Out of 140,718 New Jersey traffic incidents, 140,428 have durations of 24 hours or less.

Only 290 incidents exceed the 24-hour cutoff, so most of the dataset contains reasonable incident duration values.

These longer incidents will remain in the dataset, but they will not be used when analyzing typical incident duration.

## Step 10: Reset the Index

After filtering and cleaning the dataset, the original row numbers are no longer sequential.

The index will be reset so the cleaned traffic dataset has a simple row index starting from 0.

```python
nj_traffic = nj_traffic.reset_index(drop=True)

nj_traffic.head()
```

## Step 11: Save the Cleaned Dataset

The cleaned New Jersey traffic dataset will now be saved in the `data/processed` folder.

This processed file will later be used in the ATLAS analysis to identify traffic-risk patterns and support shipment risk monitoring.

```python
nj_traffic.to_csv(
    "../data/processed/nj_traffic_clean.csv",
    index=False
)
```
