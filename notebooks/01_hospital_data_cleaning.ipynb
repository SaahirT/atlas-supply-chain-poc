# ATLAS - Hospital Data Cleaning

## Purpose

For this notebook, the goal is to clean the CMS Hospital General Information dataset and create a smaller dataset containing only hospitals located in New Jersey.

The cleaned hospital data will later be used as possible shipment destinations in the ATLAS dashboard.

## Step 1: Import Libraries

The first step is to import the Python libraries needed for this notebook.

Pandas will be used to load, clean, filter, and save the hospital dataset.

```python
import pandas as pd
```

## Step 2: Load the Hospital Dataset

The Hospital General Information dataset comes from the Centers for Medicare & Medicaid Services (CMS).

This dataset contains information about hospitals across the United States, including their names, addresses, cities, states, ZIP codes, and other hospital-related information.

The raw dataset is stored in the `data/raw` folder.

```python
hospital_df = pd.read_csv("../data/raw/Hospital_General_Information.csv")
```

## Step 3: Inspect the Dataset

Before cleaning the data, it is important to understand what the dataset looks like.

The following cells will display:

- The first few rows
- The number of rows and columns
- The names of all the columns

```python
hospital_df.head()
```

```python
hospital_df.shape
```

```python
hospital_df.columns.tolist()
```

The dataset has 5419 rows (each representing a hospital), and 38 columns providing different information about that hospital.

## Step 4: Filter for New Jersey Hospitals

Since this is a POC for J&J, ATLAS will focus only on logistics within New Jersey.

So, the full CMS dataset will be filtered to only hospitals where the state is listed as `NJ`.

```python
nj_hospitals = hospital_df[hospital_df["State"] == "NJ"].copy()

nj_hospitals.shape
```

```python
nj_hospitals.head()
```

After filtering the dataset, we are left with 78 hospitals that are based in New Jersey.

## Step 5: Select Relevant Columns

For this project, only the columns needed to identify hospital destinations and their locations will be kept:

- Facility ID
- Facility Name
- Address
- City/Town
- State
- ZIP Code
- County/Parish

```python
columns_to_keep = [
    "Facility ID",
    "Facility Name",
    "Address",
    "City/Town",
    "State",
    "ZIP Code",
    "County/Parish"
]

nj_hospitals = nj_hospitals[columns_to_keep]

nj_hospitals.shape
```

```python
nj_hospitals.head()
```

## Step 6: Check for Missing Values

Next we will check for missing values which could cause problems later when the data is used for shipment destination or geographic analysis.

```python
nj_hospitals.isnull().sum()
```

## Step 7: Reset the Index

After filtering the original dataset to New Jersey hospitals, the original row numbers were retained.

The index will be reset so the cleaned dataset has a simple sequential index starting from 0.

```python
nj_hospitals = nj_hospitals.reset_index(drop=True)

nj_hospitals.head()
```

## Step 8: Save the Cleaned Dataset

The cleaned New Jersey hospital dataset will now be saved in the `data/processed` folder.

This processed file will be used later when ATLAS combines hospital locations with traffic-risk and shipment data.

```python
nj_hospitals.to_csv(
    "../data/processed/nj_hospitals_clean.csv",
    index=False
)
```
